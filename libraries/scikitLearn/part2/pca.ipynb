```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# dataset
data = {
    'Age':[25, 30, 35, 40, 35, 50],
    'Income':[30000, 40000, 50000, 60000, 70000, 80000],
    'Spending':[70, 60, 50, 40, 30, 20],
    'Savings':[1000, 5000, 8000, 10000, 150000, 20000]
}

df = pd.DataFrame(data)
```

```python
# Standardize data; z-score = (x - mean) / standard deviation
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
pca = PCA(n_components=2) # only 2 summarized columns
pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])
```

```python
"""
    pca.explained_variance_ratio_ -> It will tell us how much percent of total info of every
    component captured
"""

explained_variance = pca.explained_variance_ratio_
print(f"Variance captured each PCA Component")
print(np.round(explained_variance * 100, 2)) #% 95.66 0.50

"""
resume 
->  education
    work exp
    projects
    skills

    PCA1 = career readiness score
    PCA2 = special skillset score
"""
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(pca_df['PCA1'], pca_df['PCA2'], color='black', s=80)
plt.title('PCA Projection (2D View)')
plt.xlabel('PCA1 Main Pattern')
plt.ylabel('PCA2 Minor Pattern')
plt.grid(True)
plt.tight_layout()
plt.savefig('PCA_projection_2d_view.png', dpi=300, bbox_inches='tight')
plt.show()

print('New data with 2 features PCA1 PCA2')
print(pca_df)
```

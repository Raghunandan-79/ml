```python
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv("sample_data.csv")

df_label = df.copy()

le = LabelEncoder()
df_label['Gender_Encoded'] = le.fit_transform(df_label['Gender'])
df_label['Passed_Encoded'] = le.fit_transform(df_label['Passed'])

print("\nLabel Encoded")
print(df_label[['Name', 'Gender', 'Gender_Encoded', 'Passed', 'Passed_Encoded']].head())
```

```python
df_encoded = pd.get_dummies(df_label, columns=['City'], dtype=int)
print("\nOne-Hot Encoded Data (City)")
print(df_encoded)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler

data = {
    'StudyHours':[1, 2, 3, 4, 5],
    'TestScore':[40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

# standard scaler -> z = (actual value - mean) / standard deviation
standard_scaler = StandardScaler()
standard_scaled = standard_scaler.fit_transform(df)

print("Standard ScaleD Output:")
print(pd.DataFrame(standard_scaled, columns=['StudyHours', 'TestScore']))

# min max scaler -> scaled = (x - xmin) / (xmax - xmin), where x is actual value
min_max_scaler = MinMaxScaler()
min_max_scaled = min_max_scaler.fit_transform(df)
print("\nMinMax Scaled Output:")
print(pd.DataFrame(min_max_scaled, columns=['StudyHours', 'TestScore']))

x = df[['StudyHours']]
y = df[['TestScore']]

""" x_train, y_train splits 80% data for training and x_test and y_test splits 20% data for 
    testing, and random_state 42 means bring same result everytime"""
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("\nTraining Data")
print(x_train)

print("\nTest Data")
print(x_test)

print("\nTraining Data")
print(y_train)

print("\nTest Data")
print(y_test)
```

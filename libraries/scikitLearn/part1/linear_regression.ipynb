```python
from sklearn.linear_model import LinearRegression
```

```python
""" 
    [[value]] = 2d list
    we use double square brackets because model sees input as x = [[1], [2], [3], [4], [5]]
    every row is used as example and every column is used as feature

    1- its not about the line, its about the story
    2- 5 rows use start
    3- accuracy is not always goal
"""

x = [[1], [2], [3], [4], [5]]
y = [40, 50, 65, 75, 90]

model = LinearRegression()
model.fit(x, y) # x = input data, y = output data

hours = float(input("Enter how many hours you studied: "))

predicted_marks = model.predict([[hours]]) # prediction on basis of input

print(f"Based on your hours {hours} you may score around: {predicted_marks}")
```

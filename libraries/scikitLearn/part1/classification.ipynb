```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
x = [[1], [2], [3], [4], [5]] # hour studied
y = [0, 0, 0, 1, 1] # pass or fail

model = LogisticRegression()

model.fit(x, y)

hours = float(input("Enter how many hours you studied: "))
result = model.predict([[hours]])[0]

if result == 1:
    print(f"Based on hours {hours}, you are likely to PASS")
else:
    print(f"Based on hours {hours}, you are likely to FAIL")
```

```python
"""
    How KNN works
        1. Choose K value. No of neighbours. If k = 3 it means look at nearest 3 
        2. find K closest point 
        3. Count how many are in each class
        4. Predict that class which is ahead with max votes

    It works very slow for very large data
"""

x = [
    [180, 7],
    [200, 7.5],
    [250, 8],
    [300, 8.5],
    [330, 9],
    [360, 9.5]
]

# 0 = Apple, 1 = Orange
y = [0, 0, 0, 1, 1, 1]

model = KNeighborsClassifier(n_neighbors = 3)

model.fit(x, y)

weight = float(input("Enter the weight in grams: "))
size = float(input("Entee the size in cm: "))

""" 
    We are passing zero because with zero we are getting the content inside list
"""
prediction = model.predict([[weight, size]])[0]

if prediction == 0:
    print(f"This is likely an Apple.")
else:
    print(f"THis is likely an Orange.")
```

```python
"""
    Decision Tree: Machines learn on the basis of past examples
    Is fever -> 100
        Yes -> Cough ? 
            Yes -> Flu
            No -> Maybe
        No -> No flu
"""

x = [
    [7, 2], # Apple
    [8, 3], # Apple
    [9, 8], # Orange
    [10, 9] # Orange
]

y = [0, 0, 1, 1] # 0-Apple, 1-Orange

model = DecisionTreeClassifier()

model.fit(x, y)

size = float(input("Enter the fruit size in cm: "))
shade = float(input("Enter the color shade (1-10): "))

result = model.predict([[size, shade]])[0]

if result == 0:
    print(f"This is likely an apple.")
else:
    print(f"This is likely an orange.")
```

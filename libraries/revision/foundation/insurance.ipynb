# Feature Engineering and Extraction

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['bmi'])
plt.tight_layout()
plt.show()
```

```python
df_cleaned['bmi_category'] = pd.cut(
    df_cleaned['bmi'],
    bins=[0, 18.5, 24.9, 29.9, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)

df_cleaned
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['bmi_category'], drop_first=False)

df_cleaned = df_cleaned.astype(int)
df_cleaned
```

```python
df_cleaned.columns
```

```python
from sklearn.preprocessing import StandardScaler

cols = [
    'age',
    'bmi',
    'children'
]

scaler = StandardScaler()

df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])

df_cleaned.head()
```

```python
from scipy.stats import pearsonr
import pandas as pd

selected_features = [
    'age',
    'bmi',
    'children',
    'is_female',
    'is_smoker',
    'region_northeast',
    'region_northwest',
    'region_southeast',
    'region_southwest',
    'bmi_category_Underweight',
    'bmi_category_Normal',
    'bmi_category_Overweight',
    'bmi_category_Obese'
]

correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}

correlation_df = pd.DataFrame(
    list(correlations.items()),
    columns=['Feature', 'Pearson Correlation']
)

correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = [
    'is_female',
    'is_smoker',
    'region_northeast',
    'region_northwest',
    'region_southeast',
    'region_southwest',
    'bmi_category_Underweight',
    'bmi_category_Normal',
    'bmi_category_Overweight',
    'bmi_category_Obese'
]
```

```python
from scipy.stats import chi2_contingency

alpha = 0.05

df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contigency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contigency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Deciison': decision 
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python
final_df = df_cleaned[['age', 'is_female', 'bmi', 'children', 'is_smoker', 'charges', 'region_southeast', 'bmi_category_Obese']]

final_df
```
